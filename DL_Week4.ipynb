```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import SGD, RMSprop, Adam, Adagrad
import altair as alt

# Expected values (XOR problem)
y = np.array([[0, 1, 1, 0]]).T

# Features
X = np.array([[0, 0, 1, 1],
              [0, 1, 0, 1]]).T


def create_model(optimizer):
    model = Sequential()
    model.add(Dense(16, input_dim=2, activation='sigmoid'))
    model.add(Dense(1, activation='sigmoid'))

    model.compile(loss='mean_squared_error',
                  optimizer=optimizer,
                  metrics=['binary_accuracy'])
    return model


optimizers = {
    "SGD": SGD(learning_rate=0.1),
    "SGD_Momentum": SGD(learning_rate=0.1, momentum=0.9),
    "RMSprop": RMSprop(learning_rate=0.01),
    "Adam": Adam(learning_rate=0.01),
    "Adagrad": Adagrad(learning_rate=0.1)
}

results = []

for name, opt in optimizers.items():
    model = create_model(opt)
    history = model.fit(X, y, epochs=1000, verbose=0)

    for i, loss in enumerate(history.history['loss']):
        results.append({
            "Optimizer": name,
            "Epoch": i,
            "Loss": loss
        })

    # Accuracy
    y_pred = model.predict(X).round()
    accuracy = (y_pred == y).sum() / y.shape[0] * 100
    print(f"{name} Accuracy: {accuracy:.2f}%")


# Create DataFrame
df = pd.DataFrame(results)

# Plot all optimizer loss curves
chart = alt.Chart(df).mark_line().encode(
    x="Epoch",
    y="Loss",
    color="Optimizer"
).properties(
    title="Optimizer Comparison (Loss vs Epoch)"
)

chart
```
